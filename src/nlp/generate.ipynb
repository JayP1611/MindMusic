```
from transformers import GPT2Tokenizer, GPT2LMHeadModel
import torch

MODEL_PATH = "../../models/gpt2_finetuned/checkpoint-125"

tokenizer = GPT2Tokenizer.from_pretrained(MODEL_PATH)
model = GPT2LMHeadModel.from_pretrained(MODEL_PATH)
model.eval()

prompt = "<PROMPT>I fell very sad and depressed. Need to cheer myself up.<OUT>\n"

inputs = tokenizer(prompt, return_tensors = "pt") # prompt expands to input_ids and  attention_mask
```

```
with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_length = 200, # Limits the total length (input + generated tokens). Model stops once this length is reached
        temperature = 0.8, # Controls randomness of token selection
        #  | Temperature | Effect             |
        # | ----------- | ------------------ |
        # | 0.2         | very deterministic |
        # | 0.7–0.9     | balanced (best)    |
        # | 1.2+        | chaotic            |
        top_p = 0.9, # Chooses tokens from the smallest set whose cumulative probability ≥ 0.9
        do_sample = True,
        eos_token_id = tokenizer.eos_token_id
    )
```

```
text = tokenizer.decode(outputs[0], skip_special_tokens = True)
```

```
text = text.split("###")[0] + "###"
print(text)
```
